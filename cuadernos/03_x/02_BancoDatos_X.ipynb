```python
path_etiquetas = '/content/drive/MyDrive/AI-GUMELAB/CognASSISTANT-1/SEARCHES/TwitterResults/ResultadosFinalesxSerie/\
EtiquetasGumelabTwitter/tweets_IM,_FP,_CH_19.09.2022.xlsx'
```

```python
etiquetas_fp = pd.read_excel(path_etiquetas, sheet_name= "FP")
etiquetas_fp = etiquetas_fp[["Twitter ID", "Tweet"]]
etiquetas_fp = etiquetas_fp.assign(Etiqueta_FP= "FP")
```

```python
etiquetas_ch = pd.read_excel(path_etiquetas, sheet_name= "CH")
etiquetas_ch = etiquetas_ch[["Twitter ID", "Tweet"]]
etiquetas_ch = etiquetas_ch.assign(Etiqueta_CH = "CH")
etiquetas_im = pd.read_excel(path_etiquetas, sheet_name= "IM")
etiquetas_im = etiquetas_im.rename(columns={"ID tweet":'Twitter ID'})
etiquetas_im = etiquetas_im[["Twitter ID", "Tweet"]]
etiquetas_im = etiquetas_im.assign(Etiqueta_IM= "IM")
```

```python
df_fp = pd.merge(left=data, right=etiquetas_fp, how="left", left_on='text', right_on='Tweet')
df_ch = pd.merge(left=df_fp, right=etiquetas_ch, how="left", left_on='text', right_on='Tweet')
df_im = pd.merge(left=df_ch, right=etiquetas_im, how="left", left_on='text', right_on='Tweet')
```
